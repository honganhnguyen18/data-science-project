```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the language learning fluency dataset
df = pd.read_csv('language_learning_fluency.csv')

# Display basic dataset information
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
# Set up the visualization style
plt.figure(figsize=(9, 6))
sns.set_theme(style="whitegrid")

# Create a scatter plot sampling 2,000 records for clear visualization
sample_df = df.sample(n=2000, random_state=42)

sns.scatterplot(
    data=sample_df, 
    x='total_study_hours', 
    y='motivation', 
    hue='reached_fluency', 
    palette='Set1', 
    alpha=0.6,
    s=50
)

# Add titles and labels
plt.title('Total Study Hours vs. Motivation (Colored by Reached Fluency)', fontsize=14, fontweight='bold')
plt.xlabel('Total Study Hours', fontsize=12)
plt.ylabel('Motivation Score (0-10)', fontsize=12)
plt.legend(title='Reached Fluency', loc='upper right')
plt.tight_layout()

# Save the figure as required
plt.savefig('visualization1.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Set up the visualization style
plt.figure(figsize=(9, 6))

# Create a box plot showing starting age across CEFR proficiency levels
sns.boxplot(
    data=df, 
    x='cefr_level', 
    y='age_started', 
    order=['A0', 'A1', 'A2', 'B1', 'B2', 'C1', 'C2'],
    palette='Set2'
)

# Add titles and labels
plt.title('Starting Age Distribution across Attained CEFR Levels', fontsize=14, fontweight='bold')
plt.xlabel('Attained CEFR Level', fontsize=12)
plt.ylabel('Age Started Learning', fontsize=12)
plt.tight_layout()

# Save the second figure as required
plt.savefig('visualization2.png', dpi=300, bbox_inches='tight')
plt.show()
```
